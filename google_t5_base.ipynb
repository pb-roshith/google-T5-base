```python
from transformers import pipeline
```

```python
pipe = pipeline("translation", model="google-t5/t5-base")
```

```python
def tra(text, from_lang, to_lang):
  result = pipe(f"translate {from_lang} to {to_lang}: {text}.")
  print(result[0]['translation_text'])
```

```python
tra("i am roshithindia", "English", "German")
```

```python
tra("sun never sets on the British Empire", "English", "German")
```

```python
tra("i like cake", "English", "French")
```

```python
summarizer = pipeline("summarization", model="t5-base")
```

```python
def summ(text):
  input_text = "summarize: " + text
  summary = summarizer(input_text, max_length=50, min_length=20, do_sample=False)
  return summary
```

```python
summ("""
Artificial Intelligence (AI) is a rapidly growing field of computer science that focuses on building smart machines capable of performing tasks
that typically require human intelligence. These tasks include speech recognition, decision-making, visual perception, and language translation.
""")
```

```python
summ("""
With T5, we propose reframing all NLP tasks into a unified text-to-text-format where the input and output are always text strings,
in contrast to BERT-style models that can only output either a class label or a span of the input.
 Our text-to-text framework allows us to use the same model, loss function, and hyperparameters on any NLP task.
""")
```
